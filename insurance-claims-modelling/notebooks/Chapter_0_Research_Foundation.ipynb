# ── ABSTRACT ─────────────────────────────────────────────────
## Abstract

Accurate modelling of insurance claims is central to actuarial pricing,
reserving, and risk management. The classic actuarial approach separates
the claims process into two components; frequency (how many claims occur)
and severity (how much each claim costs), and combines them to estimate
the pure premium, the expected cost per unit of exposure.

This project builds and evaluates a full frequency-severity modelling
pipeline using the freMTPL2 French Motor Third Party Liability dataset,
a real world insurance dataset comprising 678,012 motor policies. The
frequency component is modelled using Poisson and Negative Binomial
regression, with the Negative Binomial included to account for
overdispersion commonly observed in claim count data. The severity
component is modelled using Gamma and Lognormal regression, both
appropriate for the right skewed, positive distribution of claim amounts.

Machine learning models — Random Forest and XGBoost are trained
alongside the GLMs to assess whether non-linear approaches offer
meaningful improvements over the interpretable actuarial baselines.
Model performance is evaluated using AIC, BIC, deviance, RMSE, MAE,
and log-likelihood. Frequency and severity predictions are combined
into a pure premium estimate, the key actuarial output of the pipeline.

**Keywords:** Insurance Claims Modelling, Frequency-Severity Framework,
Poisson Regression, Negative Binomial, Gamma Regression, Pure Premium,
XGBoost, Actuarial Pricing, freMTPL2


# ── INTRODUCTION ─────────────────────────────────────────────
## 1. Introduction

### 1.1 Background

Insurance pricing requires accurate estimation of the expected cost of
future claims. The standard actuarial approach models this expected cost
as the product of two components: claim frequency — the expected number
of claims per unit of exposure and claim severity — the expected cost
per claim. Together these produce the pure premium, the theoretical
price before expense and profit loadings.

Generalised Linear Models (GLMs) have been the foundation of actuarial
claims modelling since the seminal work of McCullagh and Nelder (1989).
Poisson regression for claim counts and Gamma regression for claim
amounts are now industry standard, embedded in regulatory frameworks
and pricing systems across the global insurance industry.

However, the emergence of machine learning has introduced powerful
alternatives. Random Forest and gradient boosting algorithms can detect
complex non-linear relationships and feature interactions that GLMs
miss, and have demonstrated superior predictive accuracy on large
insurance datasets. The question of whether this accuracy gain justifies
the loss of interpretability is a central and unresolved debate in
actuarial science.

### 1.2 Problem Statement

This project addresses three connected problems:

1. **Overdispersion in claim counts** — Real-world claim count data
   frequently exhibits overdispersion, where the variance exceeds the
   mean. Standard Poisson regression assumes variance equals mean,
   which leads to underestimated standard errors. Negative Binomial
   regression relaxes this assumption and must be tested as an
   alternative.

2. **Right-skewed claim amounts** — Claim severity distributions are
   strictly positive and right skewed, with occasional extreme values.
   The choice between Gamma and Lognormal regression affects how the
   model handles this skewness and extreme observations.

3. **GLM vs ML trade-off** — Machine learning models may offer better
   predictive accuracy but sacrifice the interpretability that actuaries
   and regulators require. This project quantifies the trade off
   empirically on a real insurance dataset.

### 1.3 Motivation

From a practical actuarial standpoint, misspecified claim models lead
directly to inadequate pricing. Underpricing creates reserving
shortfalls and potential insolvency risk. Overpricing drives away
low risk customers and reduces competitiveness. A well calibrated
frequency-severity pipeline directly supports sound actuarial pricing
decisions.

From a research standpoint, the freMTPL2 dataset provides a standardised
benchmark that allows meaningful comparison with published actuarial ML
results, strengthening the scholarly credibility of this analysis.

### 1.4 Scope

This study models claim frequency and severity separately using the
freMTPL2 frequency and severity files, merged by policy ID. The analysis
covers GLM baseline models, machine learning comparisons, and a combined
pure premium estimation. Deep learning approaches are outside the scope
of this study. All analysis is conducted in Python using open-source
libraries ensuring full reproducibility.


# ── RESEARCH QUESTIONS ────────────────────────────────────────
## 2. Research Questions and Objectives

### 2.1 Research Questions

This project is structured around four research questions:

> **RQ1:** Does the freMTPL2 claim count data exhibit overdispersion,
> and if so, does Negative Binomial regression provide a significantly
> better fit than Poisson regression as measured by AIC, BIC and
> deviance?

> **RQ2:** Between Gamma and Lognormal regression, which provides a
> better fit for claim severity in the freMTPL2 dataset, as assessed
> by log-likelihood, QQ-plots and residual diagnostics?

> **RQ3:** Do machine learning models (Random Forest and XGBoost)
> achieve meaningfully lower prediction error than GLM baselines on
> both frequency and severity, as measured by RMSE and MAE?

> **RQ4:** Which policyholder and vehicle features are the strongest
> predictors of both claim frequency and severity, and are these
> consistent across GLM and ML approaches?

### 2.2 Research Objectives

1. Merge, clean and prepare the freMTPL2 frequency and severity files
   into a single analysis ready dataset
2. Conduct exploratory data analysis including overdispersion testing,
   severity distribution fitting and correlation analysis
3. Fit and evaluate Poisson and Negative Binomial regression for
   claim frequency
4. Fit and evaluate Gamma and Lognormal regression for claim severity
5. Train and evaluate Random Forest and XGBoost on both modelling tasks
6. Compare GLM and ML performance using information criteria and
   prediction error metrics
7. Combine frequency and severity predictions into a pure premium
   estimate and validate against observed claim costs

### 2.3 Hypotheses

**H1:** The freMTPL2 claim count data will exhibit overdispersion
(variance > mean), causing Negative Binomial regression to outperform
Poisson regression on AIC and BIC.

**H2:** XGBoost will achieve lower RMSE than both GLM approaches on
severity prediction, consistent with its established advantage on
structured tabular insurance data.

**H3:** BonusMalus score and vehicle power will rank among the top
predictors for claim frequency, reflecting their actuarial relevance
as established risk factors in motor insurance.


# ── LITERATURE REVIEW ─────────────────────────────────────────
## 3. Literature Review

### 3.1 The Frequency-Severity Framework in Actuarial Science

The two-part frequency-severity model is the cornerstone of non-life
insurance pricing. Jorgensen and Paes de Souza (1994) established the
theoretical foundation for modelling claim counts with Poisson regression
and claim amounts with Gamma regression within the GLM framework.
This approach was formalised in actuarial practice by Mack (1991) and
remains the dominant methodology in pricing actuarial textbooks and
regulatory guidance.

The pure premium — the product of expected frequency and expected
severity provides the actuarially fair premium before loadings.
Frees et al. (2014) demonstrated that the two part model consistently
outperforms single equation approaches when the zero claim majority
is structurally different from the claim-filing minority.

### 3.2 Overdispersion in Claim Count Data

A well documented limitation of Poisson regression in insurance
applications is overdispersion. Cameron and Trivedi (1986) showed
that real claim count data routinely violates the Poisson assumption
of equal mean and variance, leading to underestimated standard errors
and invalid inference. The Negative Binomial distribution, which
introduces an additional dispersion parameter, was proposed as a
robust alternative. Denuit et al. (2007) confirmed that Negative
Binomial regression consistently provides better fit than Poisson on
motor insurance claim count datasets, a finding this study tests
on the freMTPL2 benchmark.

### 3.3 Severity Modelling — Gamma vs Lognormal

For claim severity, the Gamma and Lognormal distributions are the
most widely used in actuarial practice. Klugman et al. (2012)
provide a comprehensive treatment of both, noting that the Gamma
distribution is more analytically tractable within the GLM framework,
while the Lognormal may provide better tail fit for datasets with
extreme claims. Frees and Valdez (2008) compared both on US
homeowners insurance data and found the Lognormal superior in the
presence of heavy tails, a comparison this study replicates on
motor insurance data.

### 3.4 Machine Learning in Non-Life Insurance Pricing

The application of machine learning to actuarial pricing has grown
substantially. Noll et al. (2018) demonstrated that gradient boosting
models outperform GLMs on the freMTPL2 dataset for claim frequency
prediction, though the margin was smaller than expected given the
model complexity. Wüthrich (2019) introduced a neural network
framework for insurance pricing on the same dataset, establishing
freMTPL2 as a standard benchmark for actuarial ML research.

Importantly, several studies have noted that while ML models offer
higher predictive accuracy, they can produce actuarially inconsistent
premiums for example, failing to satisfy monotonicity constraints
expected by regulators. This limitation reinforces the value of
GLMs as transparent, auditable baselines.

### 3.5 The freMTPL2 Dataset as Research Benchmark

The freMTPL2 dataset has become a standard benchmark in actuarial
machine learning research. Noll et al. (2018) and Wüthrich (2019)
both used it for frequency modelling comparisons. Its combination of
678,012 policies, rich feature set, and publicly available frequency
and severity files makes it uniquely suited for a full end to end
pricing pipeline, exactly the scope of this project.

### 3.6 Research Gap

Existing studies typically address either frequency modelling or
severity modelling in isolation, or use separate datasets for each
component. This project addresses the gap by building a unified
frequency-severity pipeline on a single dataset including GLM
baselines, ML comparisons, overdispersion testing, and a combined
pure premium output within a fully reproducible Python framework.
